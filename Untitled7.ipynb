```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df3 = pd.read_csv("/content/Telco-Customer-Churn.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df3.shape)
print("\nColumns:")
print(df3.columns.tolist())
display(df3.head())

print("\nMissing values:")
print(df3.isnull().sum())
```

```python

df3.columns = df3.columns.str.strip()

# Convert TotalCharges to numeric
df3["TotalCharges"] = pd.to_numeric(
    df3["TotalCharges"].astype(str).str.strip(),
    errors="coerce"
)

# Remove customer ID because it is not a useful prediction feature
df3 = df3.drop(columns=["customerID"])

# Fill missing numeric values
df3["TotalCharges"] = df3["TotalCharges"].fillna(
    df3["TotalCharges"].median()
)

# Convert target into 0 and 1
df3["Churn"] = df3["Churn"].map({"No": 0, "Yes": 1})

print("Data cleaning completed!")
print("Missing values remaining:", df3.isnull().sum().sum())
print("\nChurn distribution:")
print(df3["Churn"].value_counts())
```

```python

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

X = df3.drop(columns=["Churn"])
y = df3["Churn"]

numeric_cols = X.select_dtypes(include=np.number).columns.tolist()
categorical_cols = X.select_dtypes(exclude=np.number).columns.tolist()

preprocessor = ColumnTransformer([
    ("numeric", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]), numeric_cols),

    ("categorical", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]), categorical_cols)
])

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Preprocessing ready!")
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, classification_report
)

models3 = {
    "Logistic Regression": Pipeline([
        ("prep", preprocessor),
        ("model", LogisticRegression(
            max_iter=2000, class_weight="balanced"
        ))
    ]),

    "Random Forest": Pipeline([
        ("prep", preprocessor),
        ("model", RandomForestClassifier(
            n_estimators=200,
            random_state=42,
            class_weight="balanced",
            n_jobs=-1
        ))
    ])
}

results3 = []
predictions3 = {}

for name, model in models3.items():
    print("Training:", name)
    model.fit(X_train, y_train)

    pred = model.predict(X_test)
    predictions3[name] = pred

    results3.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred, zero_division=0),
        "Recall": recall_score(y_test, pred, zero_division=0),
        "F1-Score": f1_score(y_test, pred, zero_division=0)
    })

results3_df = pd.DataFrame(results3)
display(results3_df.round(3))
```

```python

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for ax, (name, pred) in zip(axes, predictions3.items()):
    cm = confusion_matrix(y_test, pred)

    sns.heatmap(
        cm, annot=True, fmt="d", cmap="Blues", ax=ax,
        xticklabels=["No Churn", "Churn"],
        yticklabels=["No Churn", "Churn"]
    )

    ax.set_title(name)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")

plt.tight_layout()
plt.show()
```

```python

rf_pipeline = models3["Random Forest"]
rf_model = rf_pipeline.named_steps["model"]
feature_names = rf_pipeline.named_steps["prep"].get_feature_names_out()

importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": rf_model.feature_importances_
}).sort_values("Importance", ascending=False).head(15)

plt.figure(figsize=(10, 6))
sns.barplot(data=importance, x="Importance", y="Feature")
plt.title("Top 15 Features for Customer Churn Prediction")
plt.tight_layout()
plt.show()

display(importance)
```

```python

results3_df.to_csv("task3_churn_model_results.csv", index=False)
importance.to_csv("task3_feature_importance.csv", index=False)

print("Task 3 results saved successfully!")
print("\nClassification report:")
for name, pred in predictions3.items():
    print("\n", name)
    print(classification_report(
        y_test, pred,
        target_names=["No Churn", "Churn"],
        zero_division=0
    ))
```

```python

print("===== TASK 3 FINAL RESULTS =====")
display(results3_df.round(3))

print("\nBest model based on F1-Score:")
print(
    results3_df.loc[
        results3_df["F1-Score"].idxmax(), "Model"
    ]
)

print("\nSaved files:")
print("1. task3_churn_model_results.csv")
print("2. task3_feature_importance.csv")
```
